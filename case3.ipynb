Exercise 3 — Self-Reflection Prompt for Improving Output

Tool used: Google Colab + built-in Gemini (google.colab.ai, no API key)

PROMPT 1 — Code-generation prompt (typed into Gemini in Colab to build the notebook)
Write Python code for Google Colab that uses google.colab.ai.generate_text() (no API key) to do self-reflection in three steps: 1) summarize an article, 2) critique that summary against given criteria, 3) revise the summary based on the critique. Print the BEFORE and AFTER summaries with word counts, and a check table showing whether each one meets the length (80 words or fewer), bullet count (exactly 3), and "Key takeaway:" requirements.

PROMPT 2 — Initial summary (deliberately basic, v1)
Summarize this article.
Article: {article}

PROMPT 3 — Self-critique (explicit criteria)
You are a strict editor. Critique the summary below against the original article using these 5 criteria. Score each from 1 to 5 and give a one-line explanation:
1. Accuracy: no facts that are wrong or not in the article.
2. Completeness: covers the 3 most important points.
3. Audience: clear to a non-technical business manager (no jargon).
4. Length: 80 words or fewer.
5. Format: exactly 3 bullet points followed by a one-line "Key takeaway:".
Then list the specific changes needed to meet every criterion.
Article: {article}
Summary: {summary_v1}

PROMPT 4 — Revision
Rewrite the summary to fix every issue in the critique. The revised summary must meet all 5 criteria: accurate, covers the top 3 points, no jargon, 80 words or fewer, exactly 3 bullet points followed by one line starting with "Key takeaway:". Return only the revised summary.
Critique: {critique}
Original summary: {summary_v1}
Article: {article}

REQUIREMENTS USED FOR CRITIQUE
Accuracy, completeness (top 3 points), audience (non-technical manager), length (80 words or fewer), format (3 bullets + "Key takeaway:").

ITERATION NOTES
- The v1 prompt ("Summarize this article.") had no audience, length, or format constraints, so the first summary was longer and less structured.
- The critique prompt scored it against 5 explicit criteria, and the revision prompt required all 5 to be met.
- The BEFORE vs AFTER comparison (word count, bullets, key takeaway) is shown in the output below.


```python
from google.colab import ai

ARTICLE = """Banks and payment companies are increasingly using artificial intelligence to fight fraud, which cost consumers more than $12 billion in reported losses in the U.S. last year. Traditional fraud systems relied on fixed rules, such as flagging any purchase over a certain amount or any transaction from a new country. These rules were easy to understand but generated many false alarms, frustrating customers whose legitimate cards were declined.

Modern AI models take a different approach. Instead of fixed rules, they learn patterns from millions of past transactions, looking at signals like the time of purchase, the device used, typing speed, and how a purchase compares to the customer's normal behavior. This lets them spot unusual activity in milliseconds, before a payment is approved. Several large card networks report that AI has cut false declines by roughly a third while catching more real fraud.

However, the technology has drawbacks. AI models can be hard to explain, which is a problem when regulators or customers ask why a transaction was blocked. They also need large amounts of high-quality data, which smaller banks and credit unions may not have. And fraudsters are adapting, using AI themselves to generate fake identities and realistic phishing messages.

Experts say the best results come from combining AI with human review. Fraud analysts investigate the cases the model is unsure about, and their decisions are fed back to improve the model over time. Companies are also investing in "explainable AI" tools that show which factors drove a decision, helping them meet regulatory requirements.

Looking ahead, industry leaders expect real-time payments to grow quickly, leaving less time to stop a fraudulent transfer once it is sent. That makes fast, accurate AI detection more important than ever. For customers, the goal is simple: fewer stolen dollars and fewer embarrassing declines at the checkout counter."""

print("Article word count:", len(ARTICLE.split()))
```

```python
SUMMARY_PROMPT = "Summarize this article.\nArticle: {article}"

CRITIQUE_PROMPT = """You are a strict editor. Critique the summary below against the original article using these 5 criteria. Score each from 1 to 5 and give a one-line explanation:
1. Accuracy: no facts that are wrong or not in the article.
2. Completeness: covers the 3 most important points.
3. Audience: clear to a non-technical business manager (no jargon).
4. Length: 80 words or fewer.
5. Format: exactly 3 bullet points followed by a one-line "Key takeaway:".
Then list the specific changes needed to meet every criterion.
Article: {article}
Summary: {summary}"""

REVISE_PROMPT = """Rewrite the summary to fix every issue in the critique. The revised summary must meet all 5 criteria: accurate, covers the top 3 points, no jargon, 80 words or fewer, exactly 3 bullet points followed by one line starting with "Key takeaway:". Return only the revised summary.
Critique: {critique}
Original summary: {summary}
Article: {article}"""
```

```python
def check(text):
    words = len(text.split())
    bullets = sum(1 for l in text.splitlines() if l.strip().startswith(("-", "*", "•")))
    return {
        "words": words,
        "≤80 words": words <= 80,
        "bullets": bullets,
        "exactly 3 bullets": bullets == 3,
        "has Key takeaway": "key takeaway:" in text.lower(),
    }
```

```python
summary_v1 = ai.generate_text(SUMMARY_PROMPT.format(article=ARTICLE))
print("=" * 20, "BEFORE (v1 summary)", "=" * 20)
print(summary_v1)
```

```python
critique = ai.generate_text(CRITIQUE_PROMPT.format(article=ARTICLE, summary=summary_v1))
print("=" * 20, "SELF-CRITIQUE", "=" * 20)
print(critique)
```

```python
summary_v2 = ai.generate_text(REVISE_PROMPT.format(critique=critique, summary=summary_v1, article=ARTICLE))
print("=" * 20, "AFTER (revised summary)", "=" * 20)
print(summary_v2)
```

```python
b, a = check(summary_v1), check(summary_v2)
print("=" * 20, "BEFORE vs AFTER CHECK", "=" * 20)
print(f"{'Criterion':<20}{'Before':<12}{'After':<12}")
for k in b:
    print(f"{k:<20}{str(b[k]):<12}{str(a[k]):<12}")
```

### Iteration Notes
- **v1 prompt** ("Summarize this article.") had no audience, length, or format constraints, so the first summary was longer and less structured.
- **Self-critique prompt** scored the summary against 5 explicit criteria (accuracy, completeness, audience, length, format) and listed the specific changes needed.
- **Revision prompt** required all 5 criteria to be met.
- **Result:** The comparison table above shows the AFTER summary meets the length, bullet, and "Key takeaway" requirements the BEFORE summary lacked.
